```python
import requests

headers = {'User-Agent': 'Plant Watering Reminder App | https://github.com/Murphy96/Plant_Watering_Reminder | murphybovbjerg@gmail.com'}

zone_id = 'NC025'
start_date = '2026-08-27T23%3A00%3A00Z'
end_date = '2026-08-28T00%3A00%3A01Z'

endpoint = f'https://api.weather.gov/zones/forecast/{zone_id}/observations?start={start_date}&end={end_date}&limit=500'


response = requests.get(endpoint, headers = headers)
response.raise_for_status()
data = response.json()




#def get_weekly_weather (start_time, end_time, limit, nws_zone): 
    #response = requests.get(https://api.weather.gov/zones/forecast/CAZ043/observations?start=2026-08-20T00%3A00%3A00Z&end=2026-08-28T00%3A00%3A00Z&limit=500)  
    
```

```python
data['features']
```

```python
import pandas as pd
observations = []

for observation in data['features']:
    properties = observation['properties']

    observations.append({
        'station_id': properties['stationId'],
        'station_name': properties['stationName'],
        'timestamp': properties['timestamp'],
        'temperature': properties['temperature']['value'],
        'max_temperature_24h': properties['maxTemperatureLast24Hours']['value'],
        'min_temperature_24h': properties['minTemperatureLast24Hours']['value'], 
        'precipitation_last_3_hours': properties['precipitationLast3Hours']['value']
    })

df = pd.DataFrame(observations)

df
```
